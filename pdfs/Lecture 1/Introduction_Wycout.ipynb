# Sign Test

```python
train = []

for file in os.listdir(f"{BASE_PATH}/events"):
    if file.endswith(".json"):
        with open(f"{BASE_PATH}/events/{file}", encoding="utf-8") as f:
            data = json.load(f)
        train.append(pd.DataFrame(data))

train = pd.concat(train, ignore_index=True)
train.head()
```

```python
with open(f"{BASE_PATH}/players.json", encoding="utf-8") as f:
    players = json.load(f)

player_df = pd.DataFrame(players)
player_df.head()
```

```python
# take shots only
shots = train.loc[train["subEventName"] == "Shot"]
```

```python
son_id = player_df.loc[
    player_df["shortName"] == "Son Heung-Min",
    "wyId"
].iloc[0]
# get son's shots
son_shots = shots.loc[shots["playerId"] == son_id]
len(son_shots)
```

```python
# left leg shots
lefty_shots = son_shots.loc[
    son_shots.apply(lambda x: {'id': 401} in x.tags, axis=1)
]

# right leg shots
righty_shots = son_shots.loc[
    son_shots.apply(lambda x: {'id': 402} in x.tags, axis=1)
]

len(lefty_shots), len(righty_shots)
```

```python
# create list with +1 for left foot shots and -1 for right foot shots
l = [1] * len(lefty_shots)
l.extend([-1] * len(righty_shots))
```

```python
from statsmodels.stats.descriptivestats import sign_test

test_stat, p_value = sign_test(l, mu0=0)
test_stat, p_value
```

# T Test

```python
# 1️⃣ Mount Drive and define base path
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd, numpy as np, json, os, pathlib, warnings
warnings.filterwarnings('ignore')

BASE_PATH = "/content/drive/MyDrive/490 D/WyScout"

# 2️⃣ Load all team metadata
with open(f"{BASE_PATH}/teams.json", encoding="utf-8") as f:
    teams = json.load(f)
teams_df = pd.DataFrame(teams).rename(columns={"wyId": "teamId"})

# 3️⃣ Load events
events_df = []
for file in os.listdir(f"{BASE_PATH}/events"):
    if file.endswith(".json"):
        with open(f"{BASE_PATH}/events/{file}", encoding="utf-8") as f:
            data = json.load(f)
        events_df.append(pd.DataFrame(data))
events_df = pd.concat(events_df, ignore_index=True)

# 4️⃣ Filter Corners
corners = events_df[events_df["subEventName"] == "Corner"]

# 5️⃣ Count corners per team per match
corners_by_game = (
    corners.groupby(["teamId", "matchId"])
    .size()
    .reset_index(name="counts")
)

# 6️⃣ Merge team names
corners_by_game = corners_by_game.merge(
    teams_df[["teamId", "name"]], on="teamId", how="left"
)

# 7️⃣ One-sample t-test: Manchester City vs 6 corners
city_corners = corners_by_game.loc[
    corners_by_game["name"] == "Manchester City", "counts"
]

from scipy.stats import ttest_1samp

t_stat, p_val = ttest_1samp(city_corners, popmean=6)
print(f"One-sample t-test (City vs 6): t={t_stat:.2f}, p={p_val:.3f}")
if p_val < 0.05:
    print("Reject H0: City takes more than 6 corners per match.")
else:
    print("Cannot reject H0.")

# 8️⃣ Two-sample t-test: Liverpool vs Everton
liverpool = corners_by_game.loc[
    corners_by_game["name"] == "Liverpool", "counts"
]
everton = corners_by_game.loc[
    corners_by_game["name"] == "Everton", "counts"
]

from scipy.stats import ttest_ind

t_stat2, p_val2 = ttest_ind(liverpool, everton, equal_var=True)
print(f"Two-sample t-test: t={t_stat2:.2f}, p={p_val2:.3f}")
if p_val2 < 0.05:
    print("Reject H0: Corners differ.")
else:
    print("Cannot reject H0: No strong difference in corners.")
```
